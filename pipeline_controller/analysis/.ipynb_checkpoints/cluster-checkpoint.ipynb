```python
import psycopg2
import pandas as pd
import numpy as np
```

```python
# Definir a URL do banco de dados
DATABASE_URL = "postgresql://postgres:super#@localhost:5432/ml_social_db"

# Conectar ao banco de dados
conn = psycopg2.connect(DATABASE_URL)
cursor = conn.cursor()
```

```python
# Definir o ID para filtragem (hardcoded)
id_filtrado = 'a1789446-9083-4fdc-9a13-44568269c6ce' 

# Buscar embeddings e textos da tabela
cursor.execute("""
    SELECT embeddings, content FROM content 
    WHERE id = %s;
""", (id_filtrado,))
data = cursor.fetchall()

# Fechar conexão com o banco
cursor.close()
conn.close()
```

```python
# Converter os embeddings (BYTEA para array NumPy)
embeddings = np.array([np.frombuffer(row[0], dtype=np.float32) for row in data])
textos_originais = [row[1] for row in data]

# Verificar a dimensão dos embeddings
print(f"Shape dos embeddings: {embeddings.shape}")  # Deve ser (n_samples, n_features)
```

```python
import faiss
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist

# Função para calcular o Gap Statistic
def gap_statistic(embeddings, k_min=2, k_max=30, n_refs=30):
    gaps = []
    ks = list(range(k_min, k_max + 1))

    # Verifica se há embeddings suficientes para evitar erro
    if embeddings.shape[0] < k_max:
        k_max = embeddings.shape[0] - 1

    # Obter valores mínimo e máximo para cada dimensão
    min_vals = embeddings.min(axis=0)
    max_vals = embeddings.max(axis=0)

    # Criar distribuições de referência (dados aleatórios)
    random_references = np.array([
        np.random.uniform(low=min_vals, high=max_vals, size=embeddings.shape)
        for _ in range(n_refs)
    ])

    for k in ks:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans.fit(embeddings)
        disp = np.mean(np.min(cdist(embeddings, kmeans.cluster_centers_, "euclidean"), axis=1))

        ref_disps = []
        for j in range(n_refs):
            kmeans.fit(random_references[j])  # Treinar nos dados aleatórios
            ref_disp = np.mean(np.min(cdist(random_references[j], kmeans.cluster_centers_, "euclidean"), axis=1))
            ref_disps.append(ref_disp)

        gap = np.log(np.mean(ref_disps)) - np.log(disp)
        gaps.append(gap)

    best_k = ks[np.argmax(gaps)]
    
    return best_k, ks, gaps

# Função para encontrar o melhor k usando apenas o Gap Statistic
def encontrar_melhor_k(embeddings, k_min=2, k_max=30):
    # Garante que os embeddings sejam um array numpy
    embeddings = np.array(embeddings)

    if embeddings.shape[0] < k_min + 1:
        raise ValueError("Número insuficiente de embeddings para clusterização.")

    # 1. Gap Statistic
    best_k, ks_gap, gaps = gap_statistic(embeddings, k_min, k_max)

    # 2. Criar gráfico do Gap Statistic
    plt.figure(figsize=(6, 4))
    plt.plot(ks_gap, gaps, marker="o", linestyle="-", label="Gap Statistic")
    plt.axvline(best_k, color="r", linestyle="--", label=f"Melhor k pelo Gap = {best_k}")
    plt.xlabel("Número de Clusters (k)")
    plt.ylabel("Gap Statistic")
    plt.title("Gap Statistic para escolha do número de clusters")
    plt.legend()
    plt.show()

    print(f"🔹 Melhor número de clusters encontrado: {best_k} (Gap Statistic)")

    return best_k

# Verifica se `embeddings` está disponível
if 'embeddings' not in locals():
    raise ValueError("Variável `embeddings` não foi definida.")

best_k = encontrar_melhor_k(embeddings)

# Ajuste de `niter` e `nredo` dinamicamente baseado no tamanho dos dados
num_dados = embeddings.shape[0]
niter = 100 if num_dados < 5000 else 200  
nredo = 5 if num_dados < 5000 else 10  

# Criar clusters com FAISS usando o melhor k encontrado
kmeans = faiss.Kmeans(d=embeddings.shape[1], k=best_k, niter=niter, nredo=nredo, spherical=True)
kmeans.train(embeddings)

# Atribuir clusters
_, labels = kmeans.index.search(embeddings, 1)
labels = labels.flatten()

# Criar DataFrame para análise dos clusters
df = pd.DataFrame({"text": textos_originais, "cluster": labels})

# Exibir exemplos de cada cluster
for cluster_id in sorted(df["cluster"].unique()):
    print(f"\n🔹 **Exemplos do Cluster {cluster_id}:**")
    
    exemplos = df[df["cluster"] == cluster_id]["text"]
    if exemplos.shape[0] < 10:
        exemplos = exemplos.sample(exemplos.shape[0], random_state=42)  # Evita erro se houver poucos exemplos
    else:
        exemplos = exemplos.sample(10, random_state=42)  # Pegando 10 exemplos
    
    for idx, texto in enumerate(exemplos, start=1):
        print(f"{idx}. {texto}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import faiss

def find_best_k(embeddings, max_k=20):
    distortions = []
    k_values = range(2, max_k)
    
    for k in k_values:
        kmeans = faiss.Kmeans(d=embeddings.shape[1], k=k, niter=100, nredo=5, spherical=True)
        kmeans.train(embeddings)
        _, labels = kmeans.index.search(embeddings, 1)
        
        # Calcular inércia (soma das distâncias ao centróide mais próximo)
        centroids = kmeans.centroids
        distances = np.linalg.norm(embeddings - centroids[labels.flatten()], axis=1)
        distortions.append(np.sum(distances))
    
    # Encontrar a maior variação (cotovelo)
    diffs = np.diff(distortions)
    k_optimal = k_values[np.argmin(diffs)]  # Ponto onde a redução de inércia começa a ser menor
    
    # 🔹 Plotar o gráfico (opcional, só para validar visualmente)
    plt.plot(k_values, distortions, marker='o')
    plt.xlabel("Número de Clusters (k)")
    plt.ylabel("Inércia (Distortion)")
    plt.title("Método do Cotovelo")
    plt.show()
    
    return k_optimal

# 🔹 Encontrar o melhor k automaticamente
best_k = find_best_k(embeddings)
print(f"✅ Número ideal de clusters encontrado: {best_k}")

# 🔹 Rodar K-Means com esse `k`
kmeans = faiss.Kmeans(d=embeddings.shape[1], k=best_k, niter=300, nredo=10, spherical=True)
kmeans.train(embeddings)
_, labels = kmeans.index.search(embeddings, 1)
labels = labels.flatten()
```
