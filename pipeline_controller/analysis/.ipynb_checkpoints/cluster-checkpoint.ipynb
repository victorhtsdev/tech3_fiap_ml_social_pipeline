```python
# Converter os embeddings (BYTEA para array NumPy)
embeddings = np.array([np.frombuffer(row[0], dtype=np.float32) for row in data])
textos_originais = [row[1] for row in data]

# Verificar a dimensão dos embeddings
print(f"Shape dos embeddings: {embeddings.shape}")  # Deve ser (n_samples, n_features)
```

```python
import faiss
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist

# Função para calcular o Gap Statistic
def gap_statistic(embeddings, k_min=2, k_max=10, n_refs=10):
    gaps = []
    ks = list(range(k_min, k_max + 1))

    # Verifica se há embeddings suficientes para evitar erro
    if embeddings.shape[0] < k_max:
        k_max = embeddings.shape[0] - 1

    # Obter valores mínimo e máximo para cada dimensão
    min_vals = embeddings.min(axis=0)
    max_vals = embeddings.max(axis=0)

    # Criar distribuições de referência (dados aleatórios)
    random_references = np.array([
        np.random.uniform(low=min_vals, high=max_vals, size=embeddings.shape)
        for _ in range(n_refs)
    ])

    for k in ks:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans.fit(embeddings)
        disp = np.mean(np.min(cdist(embeddings, kmeans.cluster_centers_, "cosine"), axis=1))

        ref_disps = []
        for j in range(n_refs):
            kmeans.fit(random_references[j])  # Treinar nos dados aleatórios
            ref_disp = np.mean(np.min(cdist(random_references[j], kmeans.cluster_centers_, "cosine"), axis=1))
            ref_disps.append(ref_disp)

        gap = np.log(np.mean(ref_disps)) - np.log(disp)
        gaps.append(gap)

    best_k = ks[np.argmax(gaps)]
    
    return best_k, ks, gaps

# Função para encontrar o melhor k usando apenas o Gap Statistic
def encontrar_melhor_k(embeddings, k_min=2, k_max=20):
    # Garante que os embeddings sejam um array numpy
    embeddings = np.array(embeddings)

    if embeddings.shape[0] < k_min + 1:
        raise ValueError("Número insuficiente de embeddings para clusterização.")

    # 1. Gap Statistic
    best_k, ks_gap, gaps = gap_statistic(embeddings, k_min, k_max)

    # 2. Criar gráfico do Gap Statistic
    plt.figure(figsize=(6, 4))
    plt.plot(ks_gap, gaps, marker="o", linestyle="-", label="Gap Statistic")
    plt.axvline(best_k, color="r", linestyle="--", label=f"Melhor k pelo Gap = {best_k}")
    plt.xlabel("Número de Clusters (k)")
    plt.ylabel("Gap Statistic")
    plt.title("Gap Statistic para escolha do número de clusters")
    plt.legend()
    plt.show()

    print(f"🔹 Melhor número de clusters encontrado: {best_k} (Gap Statistic)")

    return best_k

# Verifica se `embeddings` está disponível
if 'embeddings' not in locals():
    raise ValueError("Variável `embeddings` não foi definida.")

best_k = encontrar_melhor_k(embeddings)

# Ajuste de `niter` e `nredo` dinamicamente baseado no tamanho dos dados
num_dados = embeddings.shape[0]
niter = 100 if num_dados < 5000 else 200  
nredo = 5 if num_dados < 5000 else 10  

# Criar clusters com FAISS usando o melhor k encontrado
kmeans = faiss.Kmeans(d=embeddings.shape[1], k=best_k, niter=niter, nredo=nredo, spherical=True)
kmeans.train(embeddings)

# Atribuir clusters
_, labels = kmeans.index.search(embeddings, 1)
labels = labels.flatten()

# Criar DataFrame para análise dos clusters
df = pd.DataFrame({"text": textos_originais, "cluster": labels})

# Exibir exemplos de cada cluster
for cluster_id in sorted(df["cluster"].unique()):
    print(f"\n🔹 **Exemplos do Cluster {cluster_id}:**")
    
    exemplos = df[df["cluster"] == cluster_id]["text"]
    if exemplos.shape[0] < 10:
        exemplos = exemplos.sample(exemplos.shape[0], random_state=42)  # Evita erro se houver poucos exemplos
    else:
        exemplos = exemplos.sample(10, random_state=42)  # Pegando 10 exemplos
    
    for idx, texto in enumerate(exemplos, start=1):
        print(f"{idx}. {texto}")
```

```python
import faiss
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist

# 🔹 Reduzir a dimensionalidade com PCA para analisar o número ideal de clusters
def calcular_k_pca(embeddings, variancia_min=0.95):
    pca = PCA().fit(embeddings)
    variancia_acumulada = np.cumsum(pca.explained_variance_ratio_)

    # Encontrar o número mínimo de componentes que explicam pelo menos 95% da variância
    k_pca = np.argmax(variancia_acumulada >= variancia_min) + 1

    # Aplicar a transformação PCA para reduzir dimensionalidade
    pca_embeddings = PCA(n_components=k_pca).fit_transform(embeddings)

    # Plotando a variância explicada
    plt.figure(figsize=(8,5))
    plt.plot(variancia_acumulada, marker="o")
    plt.axhline(y=variancia_min, color='r', linestyle="--", label=f"{variancia_min*100:.0f}% de variância explicada")
    plt.xlabel("Número de Componentes Principais")
    plt.ylabel("Variância Explicada Acumulada")
    plt.title("Escolha do Número de Componentes no PCA")
    plt.legend()
    plt.show()

    print(f"🔹 Número de componentes sugerido pelo PCA: {k_pca}")
    return k_pca, pca_embeddings  # Retorna também os embeddings reduzidos

# 🔹 Método do cotovelo para escolha numérica do K
def encontrar_melhor_k(embeddings, k_min=2, k_max=15):
    inercia = []
    ks = list(range(k_min, k_max + 1))

    for k in ks:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans.fit(embeddings)
        inercia.append(kmeans.inertia_)

    # Diferença de inclinação entre pontos consecutivos
    variacoes = np.diff(inercia)

    # Definição do ponto do "cotovelo"
    melhor_k = ks[np.argmin(variacoes[:k_max-2]) + 1]

    # Plotar a curva de inércia
    plt.figure(figsize=(8,5))
    plt.plot(ks, inercia, marker="o", linestyle="--")
    plt.axvline(melhor_k, color="r", linestyle="--", label=f"K escolhido = {melhor_k}")
    plt.xlabel("Número de Clusters (K)")
    plt.ylabel("Inércia (Soma das Distâncias)")
    plt.title("Método do Cotovelo para Escolha do K")
    plt.legend()
    plt.show()

    print(f"🔹 Melhor número de clusters encontrado: {melhor_k}")
    return melhor_k

# 🔹 Função para rodar FAISS com o K-Means
def rodar_kmeans_faiss(embeddings, k):
    print(f"\n🚀 Rodando K-Means com FAISS para K = {k}")
    
    # Definir FAISS KMeans
    kmeans = faiss.Kmeans(d=embeddings.shape[1], k=k, niter=300, nredo=10, spherical=True)
    kmeans.train(embeddings)

    # Atribuir clusters
    _, labels = kmeans.index.search(embeddings, 1)
    labels = labels.flatten()

    return labels

# 🔹 Função para exibir exemplos de cada cluster
def mostrar_exemplos_clusters(textos_originais, labels):
    df = pd.DataFrame({"text": textos_originais, "cluster": labels})

    for cluster_id in sorted(df["cluster"].unique()):
        print(f"\n🔹 **Exemplos do Cluster {cluster_id}:**")
        exemplos = df[df["cluster"] == cluster_id]["text"]
        if exemplos.shape[0] < 10:
            exemplos = exemplos.sample(exemplos.shape[0], random_state=42)  # Evita erro se houver poucos exemplos
        else:
            exemplos = exemplos.sample(10, random_state=42)  # Pegando 10 exemplos
        
        for idx, texto in enumerate(exemplos, start=1):
            print(f"{idx}. {texto}")

# 🔹 Pipeline de Clustering
def pipeline_clustering(embeddings, textos_originais):
    # Passo 1: Determinar K pelo PCA e reduzir dimensionalidade
    k_pca, pca_embeddings = calcular_k_pca(embeddings)

    # Passo 2: Determinar K pelo cotovelo usando os embeddings reduzidos
    k_elbow = encontrar_melhor_k(pca_embeddings)  # Usando pca_embeddings aqui

    k_final = k_elbow 

    print(f"\n🔹 Testando FAISS K-Means com K = {k_final}")

    # Rodar K-Means com FAISS nos embeddings reduzidos
    labels = rodar_kmeans_faiss(pca_embeddings, k_final)

    # Mostrar exemplos dos clusters
    mostrar_exemplos_clusters(textos_originais, labels)

# 🔹 Verifica se os embeddings estão disponíveis
if 'embeddings' not in locals():
    raise ValueError("❌ ERRO: A variável `embeddings` não foi definida. Carregue os embeddings antes de rodar.")

if 'textos_originais' not in locals():
    raise ValueError("❌ ERRO: A variável `textos_originais` não foi definida.")

# 🚀 Rodar o pipeline de clustering
pipeline_clustering(embeddings, textos_originais)
```

```python
from sklearn.mixture import GaussianMixture
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.manifold import TSNE

# Encontrar o número ideal de clusters usando BIC
bic_scores = []
k_values = range(2, 11)  # Testamos de 2 a 10 clusters

for k in k_values:
    gmm = GaussianMixture(n_components=k, covariance_type='full', random_state=42)
    gmm.fit(embeddings)
    bic_scores.append(gmm.bic(embeddings))  # Salvar o BIC

# Definir o melhor número de clusters (menor BIC)
best_k = k_values[np.argmin(bic_scores)]
print(f"Número ideal de clusters: {best_k}")

# Treinar o modelo GMM com o melhor k
gmm = GaussianMixture(n_components=best_k, covariance_type='full', random_state=42)
clusters = gmm.fit_predict(embeddings)

# Reduzir dimensionalidade para 2D com t-SNE para visualização
tsne = TSNE(n_components=2, random_state=42)
embeddings_2d = tsne.fit_transform(embeddings)

# Criar DataFrame para visualização
import pandas as pd
df_viz = pd.DataFrame({
    "x": embeddings_2d[:, 0],
    "y": embeddings_2d[:, 1],
    "cluster": clusters
})

# Plot dos clusters
plt.figure(figsize=(8, 6))
sns.scatterplot(x="x", y="y", hue="cluster", data=df_viz, palette="tab10")
plt.title("Clusters identificados com GMM")
plt.show()
```

```python
import umap
import hdbscan
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Reduzir dimensionalidade com UMAP (de embeddings originais para 2D)
umap_reducer = umap.UMAP(n_components=20, random_state=42)
embeddings_2d = umap_reducer.fit_transform(embeddings)

# Rodar HDBSCAN para encontrar clusters automaticamente
hdbscan_cluster = hdbscan.HDBSCAN(min_cluster_size=10, min_samples=3, cluster_selection_method='eom')
clusters = hdbscan_cluster.fit_predict(embeddings)

# Criar DataFrame para visualização
df_viz = pd.DataFrame({
    "x": embeddings_2d[:, 0],
    "y": embeddings_2d[:, 1],
    "cluster": clusters
})

# Plot dos clusters
plt.figure(figsize=(8, 6))
sns.scatterplot(x="x", y="y", hue="cluster", data=df_viz, palette="tab10")
plt.title("Clusters identificados com UMAP + HDBSCAN")
plt.show()

# Mostrar quantos clusters foram detectados
num_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)  # "-1" são pontos considerados ruído
print(f"Número de clusters detectados: {num_clusters}")
```
